```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_selection import SelectKBest, mutual_info_classif, chi2, mutual_info_regression

# Placeholder list of columns that need to be filled - replace with your actual columns
need_to_fill_columns = ['area', 'perimeter', 'height', 'number of floors', 'living space', 
                        'number of rooms', 'cooking equipment', '[9a]Number of residents (EFH/ MFH)', 
                #     'number of radiators', 'air duct length', 'water pipe length', 'number of toilets', 
                #    'number of sink', 'number of shower', 'number of bathtub', 'electrical cable length',
                   '[2i]Inst./Renov. heating system',
       'year of demolition of the building',
       '[9a]Number of residents (EFH/ MFH)',
       '[2h]Inst./Renov. hot water supply']

# Columns to exclude for prediction features - replace with your actual columns
exclude_columns = ['number of radiators', 'air duct length', 'water pipe length', 'number of toilets', 
                   'number of sink', 'number of shower', 'number of bathtub', 'electrical cable length']

# Specify which columns are categorical and which are numerical
categorical_columns = ['canton abbreviation', 'building status', 'building category', 'building class', 
                       'energy/heat source hot water 1', 'heat generator hot water 1', 
                       'energy/heat source heating 1', 'heat generator heating 1', 
                       'energy/heat source hot water 2', 'heat generator hot water 2', 
                       'energy/heat source heating 2', 'heat generator heating 2', 
                       '[18]Construction building', '[19]Floor plan type']

# Assuming 'df' is the DataFrame containing your data
numerical_columns = list(set(df.columns) - set(categorical_columns) - set(need_to_fill_columns) - set(exclude_columns))

feature_selection_results = {}
feature_scores = {}

for target in need_to_fill_columns:
    X = df.drop(need_to_fill_columns + exclude_columns, axis=1)  # Drop all target columns from the features
    y = df[target]  # The current target column

    # Drop rows where target is missing to ensure indices align
    not_null_indices = y.dropna().index
    X = X.loc[not_null_indices]
    y = y.loc[not_null_indices]

    # Preprocessing: Impute missing values
    num_imputer = SimpleImputer(strategy='mean')
    cat_imputer = SimpleImputer(strategy='most_frequent')

    X_num = X[numerical_columns]
    X_cat = X[categorical_columns]

    X_num_imputed = pd.DataFrame(num_imputer.fit_transform(X_num), columns=X_num.columns, index=X_num.index)
    X_cat_imputed = pd.DataFrame(cat_imputer.fit_transform(X_cat), columns=X_cat.columns, index=X_cat.index)

    # Ensure that categorical columns are of string type
    X_cat_imputed = X_cat_imputed.astype(str)

    # Preprocessing: Encode categorical variables
    encoder = OneHotEncoder(sparse_output=False)
    X_cat_encoded = encoder.fit_transform(X_cat_imputed)
    X_cat_encoded = pd.DataFrame(X_cat_encoded, columns=encoder.get_feature_names_out(X_cat.columns), index=X_cat_imputed.index)

    # Combine numerical and categorical data
    X_processed = pd.concat([X_num_imputed, X_cat_encoded], axis=1)

    # Feature Selection: Mutual Information for numerical data
    mi_selector = SelectKBest(mutual_info_regression, k='all')
    mi_selector.fit(X_num_imputed, y)
    mi_scores = mi_selector.scores_
    selected_features_mi = X_num_imputed.columns[np.argsort(mi_scores)[-7:]].tolist()  # Select top 7 features

    # Feature Selection: Chi-Squared for categorical data
    if y.dtype.kind in 'biufc':  # Ensure y is categorical or integer for chi2
        y_categorical = pd.cut(y, bins=5, labels=False)  # Binning numerical target for chi2
    else:
        y_categorical = y

    chi2_selector = SelectKBest(score_func=chi2, k=7)  # Adjust k accordingly
    chi2_selector.fit(X_cat_encoded, y_categorical)
    chi2_scores = chi2_selector.scores_
    selected_features_chi2 = [feature.split('_')[0] for feature in encoder.get_feature_names_out(X_cat.columns)[chi2_selector.get_support(indices=True)]]

    # Combine results and remove duplicates
    selected_features = list(set(selected_features_mi + selected_features_chi2))
    feature_selection_results[target] = selected_features

    # Store feature scores for visualization
    feature_scores[target] = {
        'mutual_info': dict(zip(X_num_imputed.columns, mi_scores)),
        'chi2': dict(zip(encoder.get_feature_names_out(X_cat.columns), chi2_scores))
    }

# Display the feature selection results
feature_selection_df = pd.DataFrame.from_dict(feature_selection_results, orient='index').transpose()
print(feature_selection_df)

# Aggregating feature scores
aggregated_scores = {}

for target, scores in feature_scores.items():
    for feature, score in scores['mutual_info'].items():
        if feature not in aggregated_scores:
            aggregated_scores[feature] = {'mutual_info': 0, 'chi2': 0}
        aggregated_scores[feature]['mutual_info'] += score

    for feature, score in scores['chi2'].items():
        base_feature = feature.split('_')[0]
        if base_feature not in aggregated_scores:
            aggregated_scores[base_feature] = {'mutual_info': 0, 'chi2': 0}
        aggregated_scores[base_feature]['chi2'] += score

# Visualization
mutual_info_scores = {k: v['mutual_info'] for k, v in aggregated_scores.items()}
chi2_scores = {k: v['chi2'] for k, v in aggregated_scores.items()}

# Plotting
features = list(aggregated_scores.keys())
mi_values = list(mutual_info_scores.values())
chi2_values = list(chi2_scores.values())

x = np.arange(len(features))

plt.figure(figsize=(14, 8))
plt.bar(x - 0.2, mi_values, 0.4, label='Mutual Information')
plt.bar(x + 0.2, chi2_values, 0.4, label='Chi-Squared')

plt.xlabel('Features')
plt.ylabel('Scores')
plt.title('Aggregated Feature Scores for Multiple Targets')
plt.xticks(x, features, rotation='vertical')
plt.legend()
plt.tight_layout()
plt.show()

# # Optionally save the results to a CSV file
# feature_selection_df.to_csv('feature_selection_results.csv', index=False)
```
