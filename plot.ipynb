```python
import matplotlib.pyplot as plt
import seaborn as sns

# Data for feature importance (example data similar to the image)
features = [
    "Equipment", "Apartments", "Building area", "Number of rooms",
    "Number of floors", "Height", "Reference area", "Floor", 
    "Living volume", "Type (EFH/MFH)"
]
scores = [729.0, 551.0, 504.0, 429.0, 427.0, 262.0, 144.0, 138.0, 88.0, 65.0]

# Using a Seaborn color palette
colors = sns.color_palette("Blues", len(features))

# Plotting
plt.figure(figsize=(10, 6))
plt.barh(features, scores, color=colors)
plt.xlabel('F score', fontsize=12)
plt.title('Feature importance', fontsize=14)
plt.gca().invert_yaxis()  # Invert y-axis to match the order in the example
plt.show()
```
